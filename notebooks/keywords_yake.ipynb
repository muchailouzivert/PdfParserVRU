```python
import pandas as pd
import json

df = pd.read_csv("../dataset_comparative_2020_2026/comparative_files/keywords_yake_meta.csv")
print(f"Документів: {len(df):,}")
print(df.columns.tolist())
df.head(100)
```

```python
from collections import Counter

all_kw = []
for kw_json in df["keywords_yake"]:
    try:
        kw_list = json.loads(kw_json)
        all_kw.extend([kw.lower().strip() for kw in kw_list])
    except:
        pass

counter = Counter(all_kw)
print(f"Унікальних KW: {len(counter):,}")
print(f"Всього KW: {len(all_kw):,}")
```

```python
from collections import defaultdict
import json

# Підрахунок document frequency
doc_freq = defaultdict(int)

for kw_json in df["keywords_yake"]:

    try:
        kw_list = json.loads(kw_json)

        # унікальні KW в межах одного документа
        unique_kw = set(
            kw.lower().strip()
            for kw in kw_list
        )

        for kw in unique_kw:
            doc_freq[kw] += 1

    except:
        pass

total_docs = len(df)

print(f"Всього документів: {total_docs}\n")

print(f"{'KW':<40} {'DF':>6} {'DF%':>7}")
print("-" * 60)

for kw, df_count in sorted(
    doc_freq.items(),
    key=lambda x: -x[1]
):

    pct = df_count / total_docs * 100

    print(
        f"{kw:<40} "
        f"{df_count:>6} "
        f"{pct:>6.1f}%"
    )
```

```python
from collections import defaultdict
import json

total_freq = defaultdict(int)

for kw_json in df["keywords_yake"]:

    try:
        kw_list = json.loads(kw_json)

        for kw in kw_list:

            kw = kw.lower().strip()

            if kw:
                total_freq[kw] += 1

    except:
        pass
```

```python
rubric_kw_freq = {}

for rubric, group in df.groupby("rubric"):

    temp_freq = defaultdict(int)

    for kw_json in group["keywords_yake"]:

        try:
            kw_list = json.loads(kw_json)

            # унікальні KW в межах документа
            unique_kw = set(
                kw.lower().strip()
                for kw in kw_list
            )

            for kw in unique_kw:
                temp_freq[kw] += 1

        except:
            pass

    rubric_kw_freq[rubric] = sorted(
        temp_freq.items(),
        key=lambda x: -x[1]
    )[:20]

# ВИВІД ТОП-KW ПО РУБРИКАХ

for rubric, keywords in rubric_kw_freq.items():

    print(f"\n{rubric} ")
    print(f"{'KW':<40} {'DF':>6}")
    print("-" * 50)

    for kw, count in keywords:
        print(f"{kw:<40} {count:>6}")
```

```python
DF_THRESHOLD = 0.09

stopwords_9pct = {
    kw
    for kw, cnt in doc_freq.items()
    if cnt / total_docs > DF_THRESHOLD
}

print(f"Ключових слів, які зустрічаються більше ніж у 9% документів: {len(stopwords_9pct):,}")
```

```python
removed_keywords = []

for kw, cnt in sorted(doc_freq.items(), key=lambda x: -x[1]):
    if kw in stopwords_9pct:
        removed_keywords.append({
            "keyword": kw,
            "documents": cnt,
            "percent": round(cnt / total_docs * 100, 2)
        })

removed_keywords_df = pd.DataFrame(removed_keywords)

display(removed_keywords_df)
```

```python
def filter_keywords_9pct(kw_json):
    try:
        kw_list = json.loads(kw_json)

        filtered = [
            kw.lower().strip()
            for kw in kw_list
            if kw.lower().strip() not in stopwords_9pct
        ]

        return json.dumps(filtered, ensure_ascii=False)

    except:
        return json.dumps([], ensure_ascii=False)
```

```python
df_keywords_filtered = df_original.copy()

df_keywords_filtered["keywords_yake_original"] = df_keywords_filtered["keywords_yake"]

df_keywords_filtered["keywords_yake"] = df_keywords_filtered["keywords_yake"].apply(
    filter_keywords_9pct
)

df_keywords_filtered["num_keywords_after_9pct_filter"] = df_keywords_filtered["keywords_yake"].apply(
    lambda x: len(json.loads(x)) if pd.notna(x) else 0
)

print("Середня кількість KW після фільтрації:")
print(round(df_keywords_filtered["num_keywords_after_9pct_filter"].mean(), 2))

display(df_keywords_filtered.head())
```

```python
from collections import defaultdict
import pandas as pd
import json

def get_top_keywords_by_documents(df_input, keywords_col="keywords_yake", top_n=30):
    doc_freq = defaultdict(int)
    total_docs = len(df_input)

    for kw_json in df_input[keywords_col]:
        try:
            kw_list = json.loads(kw_json)

            unique_kw = set(
                kw.lower().strip()
                for kw in kw_list
                if str(kw).strip()
            )

            for kw in unique_kw:
                doc_freq[kw] += 1

        except:
            pass

    rows = []

    for kw, cnt in sorted(doc_freq.items(), key=lambda x: -x[1])[:top_n]:
        rows.append({
            "keyword": kw,
            "documents": cnt,
            "percent_of_docs": round(cnt / total_docs * 100, 2)
        })

    return pd.DataFrame(rows)
```
